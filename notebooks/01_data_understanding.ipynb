```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/customer_shopping_behavior.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
df.info()
```

```python
print("\nMISSING VALUES")
display(
    df.isnull().sum()
)
```

```python
print("\nMISSING VALUES")
display(
    df.isnull().sum()
      .sort_values(ascending=False)
      .to_frame("Missing Values")
)
```

```python
print("\nDUPLICATE ROWS")
print("Exact duplicate rows:", df.duplicated().sum())

print("\nSUMMARY STATISTICS")
display(df.describe(include="all").T)
```

```python
print("UNIQUE VALUES PER COLUMN")
display(df.nunique().sort_values().to_frame("Unique Values"))

print("\nCATEGORY DISTRIBUTIONS")

for col in ["Gender", "Category", "Subscription Status",
            "Discount Applied", "Payment Method"]:
    print(f"\n--- {col} ---")
    display(df[col].value_counts(dropna=False))
```
