```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/customer_shopping_behavior.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
df.info()
```

```python
print("\nMISSING VALUES")
display(
    df.isnull().sum()
)
```

```python
print("\nMISSING VALUES")
display(
    df.isnull().sum()
      .sort_values(ascending=False)
      .to_frame("Missing Values")
)
```

```python
print("\nDUPLICATE ROWS")
print("Exact duplicate rows:", df.duplicated().sum())

print("\nSUMMARY STATISTICS")
display(df.describe(include="all").T)
```

```python
print("UNIQUE VALUES PER COLUMN")
display(df.nunique().sort_values().to_frame("Unique Values"))

print("\nCATEGORY DISTRIBUTIONS")

for col in ["Gender", "Category", "Subscription Status",
            "Discount Applied", "Payment Method"]:
    print(f"\n--- {col} ---")
    display(df[col].value_counts(dropna=False))
```

```python
category_analysis = (
    df.groupby("Category")
      .agg(
          Record_Count=("Customer ID", "count"),
          Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
          Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
      )
      .sort_values("Total_Purchase_Amount", ascending=False)
)

display(category_analysis.round(2))
```

```python
category_analysis = category_analysis.sort_values(
    "Total_Purchase_Amount", ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    category_analysis.index,
    category_analysis["Total_Purchase_Amount"]
)

plt.title("Total Purchase Amount by Product Category")
plt.xlabel("Total Purchase Amount (USD)")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()
```

```python
category_analysis["Contribution_%"] = (
    category_analysis["Total_Purchase_Amount"]
    / category_analysis["Total_Purchase_Amount"].sum()
) * 100

display(category_analysis.round(2))
```
